```python
Retail Sales Data Analysis
Dataset: Superstore Dataset "https://www.kaggle.com/datasets/vivek468/superstore-dataset-final"
```

```python
SELECT
    YEAR(STR_TO_DATE(`Ship Date`, '%d/%m/%Y')) AS ship_year,
    SUM(Sales) AS total_revenue
FROM superstore_sales_dataset
GROUP BY YEAR(STR_TO_DATE(`Ship Date`, '%d/%m/%Y'))
ORDER BY ship_year;
```
